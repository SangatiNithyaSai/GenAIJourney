```python
import threading
import time

def take_orders():
    for i in range(1,4):
        print("Taking order for",i)
        time.sleep(2)
    
def brew_chai():
    for i in range(1,4):
        print("Brewing chain for",i)
        time.sleep(1)
    
order_thread=threading.Thread(target=take_orders)
brew_thread=threading.Thread(target=brew_chai)

order_thread.start()
brew_thread.start()

order_thread.join()
brew_thread.join()
```

```python
from multiprocessing import Process
import time

def chai_maker(name):
    print(f"Start of {name} chai brewing")
    time.sleep(3)
    print(f"End of {name}chai brewing")

    

if __name__ =="__main__":
    chai_makers=[ Process(target=chai_maker,args=(f"Chai maker #{i+1}",)) for i in range(3)]
    start=time.time()
    for p in chai_makers:
        p.start()
    for p in chai_makers:
        p.join()
    end=time.time()

    print("All chai served by ",end-start)
```
