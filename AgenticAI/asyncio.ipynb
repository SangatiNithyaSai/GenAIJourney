```python
import nest_asyncio
nest_asyncio.apply()
import asyncio

async def brew_chai():
    print("Brewing chai")
    await asyncio.sleep(2)
    print("Chai is ready")

asyncio.run(brew_chai())
```

```python
import time
async def brew_chai(name):
    print("Brewing chai",name)
    await asyncio.sleep(2)
    print(f"{name} Chai is ready")

async def main():
    start=time.time()
    asyncio.gather(
    brew_chai("Masala"),
    brew_chai("Ginger")
    )
    print("Time take",time.time()-start)

asyncio.run(main())
```
