```python
import time
async def brew_chai(name):
    print("Brewing chai",name)
    await asyncio.sleep(2)
    print(f"{name} Chai is ready")

async def main():
    start=time.time()
    asyncio.gather(
    brew_chai("Masala"),
    brew_chai("Ginger")
    )
    print("Time take",time.time()-start)

asyncio.run(main())
```

```python
import asyncio
import aiohttp

async def fetch_url(session,url):
    async with session.get(url) as response:
        print(f"fetched {url} with status {response.status}")

async def main():
    urls=["https://httpbin.org/delay/2"]*2
    async with aiohttp.ClientSession() as session:
        tasks=[fetch_url(session,url) for url in urls]
        await asyncio.gather(*tasks)

asyncio.run(main())
```

```python
#threads with async

from concurrent.futures import ThreadPoolExecutor
import time
def check_stock(item):
    print(f"Checking {item} in store ..")
    time.sleep(3)
    return f"{item} stock :42"

async def main():
    loop=asyncio.get_running_loop()
    with ThreadPoolExecutor() as pool:
        result=await loop.run_in_executor(pool,check_stock,"Masala Chai")
        print(result)
    
asyncio.run(main())
```

```python
#Processes with async
import nest_asyncio
nest_asyncio.apply()
from concurrent.futures import ProcessPoolExecutor
import time
def encrypt(card):
    return card[::-1]

async def main():
    loop=asyncio.get_running_loop()
    with ProcessPoolExecutor() as pool:
        result=await loop.run_in_executor(pool,encrypt,"Credit_card_1234")
        print(result)
    
asyncio.run(main())
```

```python
## Background worker -> deamon=True should stop after the main function execution. But in jupyter notebook the main function will not exit,so the deamon also runs infinitely
import threading
def background_worker():
    count=1
    while True:
        time.sleep(1)
        print(f"Logging System Health-{count}\n")
        count+=1

async def fetch_orders():
    await asyncio.sleep(2)
    print("Order fetched")
threading.Thread(target=background_worker,daemon=True).start()

asyncio.run(fetch_orders())
```

```python
import ctypes
import threading

def kill_thread(thread_obj):
    if not thread_obj.is_alive():
        return
    
    # Target the internal thread ID
    target_tid = thread_obj.ident
    if target_tid is None:
        return

    # Call a low-level CPython API function to raise an exception inside that thread
    res = ctypes.pythonapi.PyThreadState_SetAsyncExc(
        ctypes.c_long(target_tid), 
        ctypes.py_object(SystemExit)
    )
    
    if res == 0:
        print(f"Failed to find thread {thread_obj.name}")
    elif res > 1:
        # If it returns > 1, something went wrong, reset it immediately
        ctypes.pythonapi.PyThreadState_SetAsyncExc(ctypes.c_long(target_tid), None)
        print("Exception injection failed.")
    else:
        print(f"Successfully terminated thread: {thread_obj.name} (ID: {target_tid})")

# Loop through all running threads and kill any thread that is NOT the MainThread
for thread in threading.enumerate():
    if thread is not threading.main_thread():
        kill_thread(thread)
```

```python
#race condition-> both try to update at the same time
import threading

chai_stock=0

def restock():
    global chai_stock
    for _ in range(100000):
        chai_stock+=1
    
threads=[ threading.Thread(target=restock) for _ in range(2)]
for t in threads: t.start()
for t in threads: t.join()


print("The final value:",chai_stock)
```

```python
#deadlock

import threading
lock_a=threading.Lock()
lock_b=threading.Lock()

def task1():
    with lock_a:
        print("Task1 acquired with lock_a")
        with lock_b:
            print("Task1 acquired with lock_b")

def task2():
    with lock_b:
        print("Task2 acquired with lock_b")
        with lock_a:
            print("Task2 acquired with lock_a")
    
t1=threading.Thread(target=task1)
t2=threading.Thread(target=task2)
t1.start()
t2.start()
```
